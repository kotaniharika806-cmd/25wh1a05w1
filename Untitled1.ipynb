```python
#Good catch — here's the full list again with a print statement added after every task so you can see the output of each one.

#Setup

#python
import os
import random
import datetime

DEVICE_TYPES = ['Temperature_Sensor', 'Humidity_Sensor', 'Pressure_Sensor', 'Motion_Sensor']
LOCATIONS = ['Warehouse_A', 'Warehouse_B', 'Office_1', 'Office_2', 'Server_Room']
READING_RANGES = {
    'Temperature_Sensor': (15.0, 35.0),
    'Humidity_Sensor': (20.0, 90.0),
    'Pressure_Sensor': (980.0, 1050.0),
    'Motion_Sensor': (0.0, 1.0)
}

#1.1 — Generate 100 simulated sensor readings

#python
raw_sensor_data = []
start_time = datetime.datetime(2025, 1, 1, 0, 0, 0)
for i in range(1, 101):
    timestamp_str = (start_time + datetime.timedelta(minutes=i * 15)).strftime('%Y-%m-%d %H:%M:%S')
    device_id = f"DEV{i:03d}"
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)
    low, high = READING_RANGES[device_type]
    value = round(random.uniform(low, high), 2)
    raw_sensor_data.append((timestamp_str, device_id, device_type, location, value))

print(f"Task 1.1 -> Generated {len(raw_sensor_data)} readings. First reading: {raw_sensor_data[0]}")

#1.2 — Extract temperature readings

#python
temperature_readings = []
for reading in raw_sensor_data:
    timestamp, device_id, device_type, location, value = reading
    if device_type == 'Temperature_Sensor':
        temperature_readings.append({
            'timestamp': timestamp,
            'device_id': device_id,
            'location': location,
            'temperature_celsius': value
        })

print(f"Task 1.2 -> {len(temperature_readings)} temperature readings. First one: {temperature_readings[0]}")

#1.3 — Unique locations

#python
unique_locations = {reading[3] for reading in raw_sensor_data}

print(f"Task 1.3 -> unique_locations: {unique_locations}")

#1.4 — Device stats (min, max) per device_type

#python
device_stats = {}
for device_type in DEVICE_TYPES:
    device_stats[device_type] = (float('inf'), float('-inf'))

for reading in raw_sensor_data:
    device_type = reading[2]
    value = reading[4]
    current_min, current_max = device_stats[device_type]
    device_stats[device_type] = (min(current_min, value), max(current_max, value))

print(f"Task 1.4 -> device_stats: {device_stats}")

#1.5.1 — List slicing

#python
sliced_readings_1 = raw_sensor_data[9:21]   # 10th to 20th reading (inclusive)
sliced_readings_2 = raw_sensor_data[::5]    # every 5th reading

print(f"Task 1.5.1 -> sliced_readings_1 ({len(sliced_readings_1)} items): {sliced_readings_1}")
print(f"Task 1.5.1 -> sliced_readings_2 ({len(sliced_readings_2)} items): {sliced_readings_2}")

#1.5.2 — String slicing

#python
sample_string = "PythonProgrammingIsFun"

substring_1 = sample_string[6:17]   # "Programming"
substring_2 = sample_string[5::-1]  # "nohtyP"

print(f"Task 1.5.2 -> substring_1: {substring_1}")
print(f"Task 1.5.2 -> substring_2: {substring_2}")

#1.5.3 — Mixed slicing

#python
fifth_reading_device_type = raw_sensor_data[4][2]
device_type_suffix = fifth_reading_device_type[-6:]

print(f"Task 1.5.3 -> fifth_reading_device_type: {fifth_reading_device_type}")
print(f"Task 1.5.3 -> device_type_suffix: {device_type_suffix}")

#2.1 — get_readings_by_location()

#python
def get_readings_by_location(readings_list, target_location):
    return [reading for reading in readings_list if reading[3] == target_location]

sample_location = raw_sensor_data[0][3]
result_2_1 = get_readings_by_location(raw_sensor_data, sample_location)
print(f"Task 2.1 -> {len(result_2_1)} readings found for location '{sample_location}'")

#2.2 — calculate_average_temperature()

#python
def calculate_average_temperature(temp_data_list):
    if not temp_data_list:
        return 0.0
    total = sum(item['temperature_celsius'] for item in temp_data_list)
    return total / len(temp_data_list)

average_temp = calculate_average_temperature(temperature_readings)
print(f"Task 2.2 -> average temperature: {average_temp:.2f}°C")

#2.3 — generate_report_line()

#python
def generate_report_line(reading_dict):
    return (f"[{reading_dict['timestamp']}] Device {reading_dict['device_id']} "
            f"at {reading_dict['location']}: {reading_dict['temperature_celsius']}°C")

print(f"Task 2.3 -> {generate_report_line(temperature_readings[0])}")

#3.1 — Create 'sensor_reports' directory

#python
reports_dir = 'sensor_reports'
if not os.path.exists(reports_dir):
    os.makedirs(reports_dir)
    print(f"Task 3.1 -> Directory '{reports_dir}' created.")
else:
    print(f"Task 3.1 -> Directory '{reports_dir}' already exists.")

#3.2 — Duration between first and last timestamp

#python
if len(raw_sensor_data) > 1:
    first_timestamp_str = raw_sensor_data[0][0]
    last_timestamp_str = raw_sensor_data[-1][0]

    date_format = '%Y-%m-%d %H:%M:%S'
    first_dt = datetime.datetime.strptime(first_timestamp_str, date_format)
    last_dt = datetime.datetime.strptime(last_timestamp_str, date_format)

    duration = last_dt - first_dt

    days = duration.days
    seconds = duration.total_seconds()
    hours = int(seconds // 3600) % 24
    minutes = int((seconds % 3600) // 60)

    print(f"Task 3.2 -> Duration: {days} days, {hours} hours, {minutes} minutes.")
else:
    print("Task 3.2 -> Not enough data to calculate duration.")

#3.3 — 5 random integers between 100–200

#python
random_values = [random.randint(100, 200) for _ in range(5)]
print(f"Task 3.3 -> random_values: {random_values}")

#4.1 — Write temperature readings to file

#python
report_filename = os.path.join(reports_dir, 'temperature_report.txt')
try:
    with open(report_filename, 'w') as f:
        for reading_dict in temperature_readings:
            line = generate_report_line(reading_dict)
            f.write(line + '\n')
    print(f"Task 4.1 -> Report written to '{report_filename}'.")
except IOError as e:
    print(f"Task 4.1 -> Error writing file '{report_filename}': {e}")

#4.2 — Read the file back

#python
read_report_lines = []
try:
    with open(report_filename, 'r') as f:
        read_report_lines = f.readlines()
    print(f"Task 4.2 -> Read {len(read_report_lines)} lines. First 3:")
    for line in read_report_lines[:3]:
        print(line.strip())
except IOError as e:
    print(f"Task 4.2 -> Error reading file '{report_filename}': {e}")
```
