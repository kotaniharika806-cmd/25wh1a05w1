```python
import numpy as np
import numpy.linalg as la          # inv, norm, eig, solve, matrix_rank
import matplotlib.pyplot as plt

np.set_printoptions(precision=3, suppress=True)   # tidy array printing
np.random.seed(42)
print('Setup complete. NumPy', np.__version__)
A = np.array([[2., 0.],
              [0., 3.]])

vals, vecs = la.eig(A)     # vals = eigenvalues, vecs columns = eigenvectors
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)


# -----------------------------------------------------------
# 🔹 5B. VERIFY  A v = lambda v
# -----------------------------------------------------------

for i in range(len(vals)):
    v = vecs[:, i]                 # i-th eigenvector
    lhs = A @ v                    # A v
    rhs = vals[i] * v              # lambda v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
C = np.array([[4., 1.],
              [2., 3.]])
vals_C, vecs_C = la.eig(C)

print('\nMatrix C:\n', C)
print('Eigenvalues of C:', vals_C)
print('Eigenvectors of C (columns):\n', vecs_C)


# 2. Verify C v = lambda v
for i in range(len(vals_C)):
    v = vecs_C[:, i]

    lhs = C @ v
    rhs = vals_C[i] * v

    print(f'lambda={vals_C[i]:.1f}  C v == lambda v ?',
          np.allclose(lhs, rhs))


# 3. Print each eigenvalue with its eigenvector
print('\nEigenvalue-eigenvector pairs:')

for i in range(len(vals_C)):
    print('lambda =', vals_C[i])
    print('v      =', vecs_C[:, i])


# 4. Verify using the eigenvalue equation
#    C @ V = V @ Lambda
Lambda = np.diag(vals_C)

print('\nC @ V == V @ Lambda ?',
      np.allclose(C @ vecs_C, vecs_C @ Lambda))
```
